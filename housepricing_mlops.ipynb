```python
import pandas as pd
import numpy as np

# Load Boston dataset manually
url = "http://lib.stat.cmu.edu/datasets/boston"
raw_df = pd.read_csv(url, sep=r"\s+", skiprows=22, header=None)

X = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
y = raw_df.values[1::2, 2]

# Define feature names (same as original sklearn)
feature_names = [
    "CRIM","ZN","INDUS","CHAS","NOX","RM","AGE",
    "DIS","RAD","TAX","PTRATIO","B","LSTAT"
]

# ✅ Create boston object (like sklearn used to give)
class Boston:
    pass

boston = Boston()
boston.data = X
boston.target = y
boston.feature_names = feature_names
boston.DESCR = "Boston Housing Dataset (manually loaded)"

# Optional: dataframe
boston.df = pd.DataFrame(X, columns=feature_names)
boston.df["PRICE"] = y
```

```python
from  sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
#
```

# Model Trainig

```python
from sklearn.linear_model import LinearRegression
#from sklearn.linear_model import Lasso
#from sklearn.linear_model import Ridge
#from sklearn.linear_model import ElasticNet
#from sklearn.metrics import accuracy_score,r2_score,mean_absolute_error
```

```python
regression=LinearRegression()
```

```python
regression.fit(X_train,y_train)
```

```python
# print the coefficients and the intercept
print(regression.coef_)
```

```python
print(regression.intercept_)
```

```python
## on which parameters the model has been trained
regression.get_params()
```

```python
### Prediction With Test Data
reg_pred=regression.predict(X_test)
```

```python
reg_pred
```

#Assumptions

```python
## plot a scatter plot for the prediction
plt.scatter(y_test,reg_pred)
```

```python
## Residuals
residuals=y_test-reg_pred
```

```python
residuals
```

```python
## plot  this residuals
sns.displot(residuals,kind="kde")
```

```python
## Scatter plot with respect to prediction and residuals
## uniform distribution
plt.scatter(reg_pred,residuals)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error

print(mean_absolute_error(y_test,reg_pred))
print(mean_squared_error(y_test,reg_pred))
print(np.sqrt(mean_squared_error(y_test,reg_pred)))
```

# R square and adjusted R square

Formula

R^2=1 SSR/SST

R^2=Cofficient of determination SSR SST

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,reg_pred)
print(score)
```

Ajusted R2=1-[(1-R2)*(n-1)/(n-k-1)]
where :
R2: The R2 of the model n : The number of observation k: The number of predictore variables

```python
#display adjusted R-squared
1 - (1-score)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```
